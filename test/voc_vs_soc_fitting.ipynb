```python
# Read a mat file and extract the data into a dataframe
from h11 import Data
import scipy
import pandas as pd


def read_mat_file(file_path):
    mat_data = scipy.io.loadmat(file_path)
    meas = mat_data['meas']

    # here are the columns within the mat file:
    # Data columns:
    # TimeStamp (timestamp in MM/DD/YYYY HH:MM:SS AM format)
    # Voltage (measured cell terminal voltage, sense leads welded directly to battery terminal)
    # Current (measure current in amps)
    # Ah (measured amp-hours, with Ah counter reset before each charge, test, or drive cycle)
    # Wh (measured watt-hours, with Wh counter reset after each charge, test, or drive cycle)
    # Power (measure power in watts)
    # Battery_Temp_degC (battery case temperature, at middle of battery, in degrees Celsius measured with a thermocouple, measurement corrected for thermocouple offset at lower temperatures)
    # Time (time in seconds, starts at zero at beginning of each data file)
    # Chamber_Temp_degC (measured chamber temperature in degrees Celsius)    

    # A list of names for all the columns in the mat file:

    columns = ['TimeStamp', 'Voltage', 'Current', 'Ah', 'Wh', 'Power', 'Battery_Temp_degC', 'Time', 'Chamber_Temp_degC']

    print(meas.dtype)  # Check the data type of the loaded mat file

    # a dataframe with the data from the mat file, with the columns named according to the list above:
    data_df = pd.DataFrame(columns=columns)
        
    for col in columns:
        if col in data_df.columns:
            # A special case for the TimeStamp column, which is stored as a structured array in the mat file:
            if col == 'TimeStamp':
                data_df[col] = [str(ts[0]) for ts in meas[col][0, 0].flatten()]  # Convert structured array to list of strings
            else:
            
                data_df[col] = meas[col][0, 0].flatten()  # Flatten the column to make it 1D
        else:
            print(f"Warning: Column '{col}' not found in the mat file.")
            data_df[col] = None  # Fill missing columns with None    

    # print the dataframe to check that it was read correctly:
    print(data_df)
       
    
    return data_df  
    
```

```python
# the mat file name 
import os


file_name = "05-08-17_13.26 C20 OCV Test_C20_25dC"

mat_file_name = os.path.join("data", file_name+".mat")  
# Read the mat file and extract the data into a dataframe
data_df = read_mat_file(mat_file_name)

# save the dataframe to a csv file
csv_file_name = os.path.join("data", file_name+".csv")
data_df.to_csv(csv_file_name, index=False)
```
